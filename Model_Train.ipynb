```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
print("GPU available:", len(tf.config.list_physical_devices("GPU")) > 0)
```

```python

import tensorflow as tf
from tensorflow.keras import layers, models

IMG_SIZE = 224
NUM_CLASSES = len(dataset.features["label"].names)

# Load the pretrained MobileNetV2 model
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze the pretrained layers initially
base_model.trainable = False

# Define the input layer
inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))

# Preprocess images for MobileNetV2
x = tf.keras.applications.mobilenet_v2.preprocess_input(inputs)

# Extract image features using the pretrained model
x = base_model(x, training=False)

# Reduce the spatial dimensions
x = layers.GlobalAveragePooling2D()(x)

# Apply dropout to help reduce overfitting
x = layers.Dropout(0.2)(x)

# Add the output layer for 80 plant classes
outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)

# Create the final classification model
model = models.Model(inputs, outputs)

# Compile the model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Display the model architecture
model.summary()
```

```python
IMG_SIZE = 224
BATCH_SIZE = 16

def make_tf_dataset(hf_dataset, training=False):
    def image_generator():
        for item in hf_dataset:
            image = item["image"].convert("RGB")
            image = image.resize((IMG_SIZE, IMG_SIZE))
            image = np.asarray(image, dtype=np.float32)

            label = np.int64(item["label"])
            yield image, label

    output_signature = (
        tf.TensorSpec(
            shape=(IMG_SIZE, IMG_SIZE, 3),
            dtype=tf.float32
        ),
        tf.TensorSpec(shape=(), dtype=tf.int64)
    )

    tf_dataset = tf.data.Dataset.from_generator(
        image_generator,
        output_signature=output_signature
    )

    if training:
        tf_dataset = tf_dataset.shuffle(
            buffer_size=1000,
            seed=42
        )

    tf_dataset = tf_dataset.batch(BATCH_SIZE)
    tf_dataset = tf_dataset.prefetch(tf.data.AUTOTUNE)

    return tf_dataset

train_ds = make_tf_dataset(train_data, training=True)
val_ds = make_tf_dataset(val_data)
test_ds = make_tf_dataset(test_data)

print("Training, validation, and testing datasets are ready.")
```

```python

EPOCHS = 5

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)
```

```python
test_loss, test_accuracy = model.evaluate(test_ds)

print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
print(f"Test Loss: {test_loss:.4f}")
```

```python
from sklearn.metrics import classification_report

# Collect actual labels and model predictions
y_true = []
y_pred = []

for images, labels in test_ds:
    predictions = model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(predictions, axis=1))

# Get class names from the dataset
class_names = dataset.features["label"].names

# Generate the classification report
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(len(class_names))),
    target_names=class_names,
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Calculate the confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(len(class_names)))
)

# Create the confusion matrix plot
fig, ax = plt.subplots(figsize=(22, 22))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    ax=ax,
    xticks_rotation="vertical",
    include_values=False,
    cmap="Blues",
    colorbar=False
)

plt.title("Confusion Matrix - Herbal Plant Classification")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Load the herbal knowledge base
herbs_df = pd.read_csv("/content/herbs.csv").fillna("")

def get_plant_information(plant_name):
    # Find the matching plant
    match = herbs_df[
        herbs_df["class_label"].str.strip().str.casefold()
        == plant_name.strip().casefold()
    ]

    if match.empty:
        return {"message": "Plant information not found."}

    # Return only available information
    info = match.iloc[0].to_dict()

    return {
        key: value
        for key, value in info.items()
        if str(value).strip()
    }

# Test the function
result = get_plant_information("Tulsi")
print(result)
```

```python
import numpy as np
from PIL import Image

def predict_plant_with_information(image_path):
    # Load and prepare the image
    img = Image.open(image_path).convert("RGB")
    img = img.resize((224, 224))

    # Convert image to an array
    img_array = np.array(img, dtype=np.float32)
    img_array = np.expand_dims(img_array, axis=0)

    # Predict the plant
    predictions = model.predict(img_array, verbose=0)
    predicted_index = int(np.argmax(predictions[0]))
    confidence = float(predictions[0][predicted_index]) * 100

    # Get the predicted plant label
    plant_name = class_names[predicted_index]

    # Fetch matching herbal information
    information = get_plant_information(plant_name)

    return {
        "predicted_plant": plant_name,
        "confidence_percent": round(confidence, 2),
        "herbal_information": information
    }

print("Prediction and herbal lookup function created!")
```

```python
import tensorflow as tf
import numpy as np
import json
from PIL import Image

# Load the trained model
model = tf.keras.models.load_model("ayurvedic_plant_model.keras")

# Load plant class names
with open("plant_class_names.json", "r") as f:
    class_names = json.load(f)

# Define the prediction function
def predict_plant(image_path):
    # Load and preprocess the image
    image = Image.open(image_path).convert("RGB")
    image = image.resize((224, 224))

    image_array = np.array(image, dtype=np.float32)
    image_array = np.expand_dims(image_array, axis=0)

    # Generate predictions
    predictions = model.predict(image_array, verbose=0)
    predicted_index = int(np.argmax(predictions[0]))

    # Get the predicted plant name and confidence
    plant_name = class_names[predicted_index]
    confidence = float(predictions[0][predicted_index]) * 100

    return {
        "plant_name": plant_name,
        "confidence": round(confidence, 2)
    }

print("Prediction function is ready!")
```

```python
from fastapi import FastAPI, File, UploadFile, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from PIL import UnidentifiedImageError

app = FastAPI(
    title="AI-Based Ayurvedic Herbal Knowledge API",
    version="1.0.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"]
)

@app.get("/")
def home():
    return {"message": "Ayurvedic Plant API is ready", "status": "running"}

@app.get("/plants")
def list_plants():
    return {"total_plants": len(class_names), "plants": class_names}

@app.post("/predict")
async def predict(file: UploadFile = File(...)):
    if file.content_type not in ["image/jpeg", "image/png", "image/webp"]:
        raise HTTPException(400, "Upload a JPG, PNG, or WEBP image.")

    contents = await file.read()

    if len(contents) > 10 * 1024 * 1024:
        raise HTTPException(413, "Image must be 10 MB or smaller.")

    try:
        image = Image.open(io.BytesIO(contents)).convert("RGB")
        image = image.resize((224, 224))
    except (UnidentifiedImageError, OSError):
        raise HTTPException(400, "Invalid image file.")

    image_array = np.array(image, dtype=np.float32)[None, ...]
    predictions = model.predict(image_array, verbose=0)[0]

    index = int(np.argmax(predictions))
    plant_name = class_names[index]

    knowledge = plant_knowledge.get(plant_name, {
        "common_name": plant_name,
        "scientific_name": "Not verified",
        "traditional_uses": [
            "Information is not available in the database yet."
        ],
        "note": "AI identification may be incorrect."
    })

    return {
        "predicted_plant": plant_name,
        "confidence": round(float(predictions[index]) * 100, 2),
        "plant_information": knowledge,
        "disclaimer": "AI prediction only; not confirmed identification or medical advice."
    }

print("API endpoints created.")
```

```python
%%writefile /content/app.py
import streamlit as st
import requests
from PIL import Image

API_URL = "https://heatedly-relapsable-graeme.ngrok-free.dev"

st.set_page_config(
    page_title="Ayurvedic Herbal Knowledge System",
    page_icon="🌿",
    layout="centered"
)

st.title("🌿 AI-Based Ayurvedic Herbal Knowledge System")
st.write("Upload a medicinal plant image to identify the plant.")

uploaded_file = st.file_uploader(
    "Choose a plant image",
    type=["jpg", "jpeg", "png"]
)

if uploaded_file is not None:
    image = Image.open(uploaded_file).convert("RGB")
    st.image(image, caption="Uploaded Plant Image", use_container_width=True)

    if st.button("Predict Plant"):
        with st.spinner("Identifying plant..."):
            try:
                uploaded_file.seek(0)

                response = requests.post(
                    f"{API_URL}/predict",
                    files={
                        "file": (
                            uploaded_file.name,
                            uploaded_file.getvalue(),
                            uploaded_file.type
                        )
                    },
                    timeout=120
                )

                if response.status_code == 200:
                    result = response.json()

                    st.subheader("🌱 Prediction Result")
                    st.success(result["predicted_plant"])
                    st.metric("Confidence", f'{result["confidence"]}%')

                    info = result.get("plant_information", {})
                    st.subheader("Herbal Information")

                    st.write("**Common Name:**", info.get("common_name", "Not available"))
                    st.write("**Scientific Name:**", info.get("scientific_name", "Not available"))

                    st.write("**Traditional Uses:**")
                    for use in info.get("traditional_uses", []):
                        st.write(f"- {use}")

                    st.info(info.get("note", "Traditional information only."))

                    st.warning(result.get(
                        "disclaimer",
                        "AI prediction is not confirmed identification or medical advice."
                    ))
                else:
                    st.error(f"API error: {response.status_code}")
                    st.write(response.text)

            except requests.exceptions.RequestException as error:
                st.error("Could not connect to the prediction API.")
                st.write(str(error))
```
